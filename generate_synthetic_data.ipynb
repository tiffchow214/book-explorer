```python
import pandas as pd
import numpy as np
from faker import Faker
import datetime
from IPython.display import display

def create_synthetic_data(num_rows=1000, show=True, head=None, seed=42):
    rng = np.random.default_rng(seed)
    fake = Faker('en_US')

    start_date = datetime.date(2022, 1, 1)
    end_date = datetime.date.today()
    regions = ['North', 'South', 'East', 'West']

    rows = []
    for _ in range(num_rows):
        cust_id = fake.uuid4()
        sdate = fake.date_between(start_date=start_date, end_date=end_date)
        reg = rng.choice(regions)

        # slightly more realistic distributions
        ages = int(np.clip(rng.normal(45, 12), 18, 85))
        purchase_freq = int(np.clip(rng.poisson(8), 1, 30))
        last_login_days = int(np.clip(rng.integers(0, 365), 0, 365))
        total_spent = int(np.clip(rng.normal(40 * purchase_freq, 80), 0, 5000))

        # churn logic using the current row's values
        if purchase_freq < 5 and last_login_days > 50:
            churn = rng.choice([1, 0], p=[0.8, 0.2])
        else:
            churn = rng.choice([1, 0], p=[0.1, 0.9])

        rows.append((
            cust_id, sdate, reg, ages, purchase_freq, last_login_days, total_spent, churn
        ))

    df = pd.DataFrame(rows, columns=[
        'customer_id', 'signup_date', 'region', 'age',
        'purchase_frequency', 'last_login_days', 'total_spent', 'churn_status'
    ])

    # Make sure dates are proper dtype and columns order/width render nicely in notebooks
    df['signup_date'] = pd.to_datetime(df['signup_date'])

    # Notebook-friendly display (rich HTML). Shows the head if requested.
    if show:
        display(df.head(head) if head else df)

    return df

synthetic_df = create_synthetic_data(1000, show=True, head=10)
```
